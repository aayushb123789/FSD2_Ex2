```python
# Importing the pandas library
import pandas as pd
```

```python
# Reading the csv file with no header
data = pd.read_csv("retail_sales.csv", header=None)
```

```python
# Assigning column names
data.columns = ['Order_ID', 'Product', 'Category', 'Region', 'Month', 'Sales', 'Quantity', 'Profit']
# Displaying the first five records
data.head()
```

```python
# Displaying the missing value frequency in each column
print(data.isnull().sum())
```

```python
# Filling the missing values
data['Product'].fillna(data['Product'].mode()[0], inplace=True)
data['Sales'].fillna(data['Sales'].mean(), inplace=True)
data['Profit'].fillna(data['Profit'].mean(), inplace=True)
```

```python
# Displaying the missing value frequency in each column after filling missing value
print(data.isnull().sum())
```

```python
# Calculating total Sales
total_sales = data['Sales'].sum()
print("Total Sales:", total_sales)
```

```python
# Calculating total Profit
total_profit = data['Profit'].sum()
print("Total Profit:", total_profit)
```

```python
revenue_region = data.groupby('Region')['Sales'].sum()
print(revenue_region)
```

```python
highest_region=revenue_region.idxmax()
print("Region with Highest Revenue:", highest_region)
```

```python
monthly_sales = data.groupby('Month')['Sales'].sum()
print("Monthly Sales:", monthly_sales)
```

```python
peak_month = monthly_sales.idxmax()
print("Peak Sales Month:", peak_month)
```

```python
import matplotlib.pyplot as plt

monthly_sales = data.groupby('Month')['Sales'].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='royalblue')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Profit', data=data, estimator=sum, palette='viridis')
plt.title('Category-wise Profit Distribution')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.show()
```
